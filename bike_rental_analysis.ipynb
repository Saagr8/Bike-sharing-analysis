```python
import matplotlib.pyplot as plt
```

```python
working_avg=df_h[df_h["workingday"]==1].groupby("hr")["cnt"].mean()
```

```python
non_working_avg=df_h[df_h["workingday"]==0].groupby("hr")["cnt"].mean()
```

```python
plt.plot(working_avg.index,working_avg.values,label="Working Days")
plt.plot(non_working_avg.index,non_working_avg.values,label="Non Working Day")
plt.title("Working Day vs Non Working Day")
plt.ylabel("Average Bike Rentals")
plt.xlabel("Hour of the day (24h-Format)")
plt.legend()
```

```python
working_hr=df_h[(df_h["workingday"]==1) & (df_h["hr"]==17)]
```

```python
all_correct = True

for i in working_hr["hr"]:
    if i != 17:
        all_correct = False
        break

if all_correct and not working_hr.empty:
    print("All hours are 17")
else:
    print("Data needs checking")
```

```python

low_wind = working_hr[working_hr["windspeed"] < 0.4]
high_wind = working_hr[working_hr["windspeed"] >= 0.4]

plt.figure(figsize=(9, 5))

plt.scatter(low_wind["windspeed"], low_wind["cnt"],
            color="royalblue", alpha=0.4, marker="o",
            label="Wind speed below 0.4")

plt.scatter(high_wind["windspeed"], high_wind["cnt"],
            color="orange", alpha=0.7, marker="x",
            label="Wind speed 0.4 or above")

plt.xlabel("Normalised Wind Speed")
plt.ylabel("Bike Rentals")
plt.title("Wind Speed vs Bike Rentals at 5 PM on Working Days")
plt.legend()
plt.show()
```

```python
## Key Findings and Limitations

# In this project, I analysed bike rental data to understand when people rent bikes the most and how weather conditions affect rentals.

# I found that on working days, bike rentals are higher around 8 AM and 5–6 PM. I think this could be because people are going to work and coming back home. On non-working days, the busiest time was around 1 PM.

# I also noticed that bike rentals are generally higher during good weather and lower during bad weather. However, the worst weather category only had 3 observations, so I cannot fully depend on that result.

# When I checked wind speed, I couldn't find a very clear relationship with bike rentals. I think other factors, like time of day and weather conditions, also play a role.

# Overall, this project helped me understand how to explore data, compare different conditions and find patterns using Python.
```
